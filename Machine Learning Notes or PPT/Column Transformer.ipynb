### **What is ColumnTransformer?**

ColumnTransformer (from scikit-learn) is a tool that helps you:

 Apply different preprocessing techniques to different columns of your dataset — all in one place.
 ____________________________________________________________________________

 **ColumnTransformer**

 Apply different preprocessing techniques to different columns in a dataset efficiently.


### **Why do we need it?**

In real-world datasets:

Some columns are numerical → need scaling (e.g., StandardScaler)
Some are categorical → need encoding (e.g., OneHotEncoder)

Instead of handling them separately, ColumnTransformer does everything together in a pipeline.

**Importing useful library**

```python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder
```

```python
from google.colab import files

file = files.upload()
```

```python
df = pd.read_csv("covid_toy.csv")
```

```python
df.head()
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
df2 = transformer.fit_transform(df)
```

### **Proper Data Preprocessing**
